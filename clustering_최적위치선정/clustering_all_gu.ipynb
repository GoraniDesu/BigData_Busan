# Meanshift로 구현

```python
# MeanShift 구현
import folium
from sklearn.cluster import MeanShift
import matplotlib.pyplot as plt
from shapely.geometry import Point

# MeanShift의 대역폭(bandwidth) 설정
# 이 값은 5~9세 아동의 15분 평균 보행 거리(약 540m)를 기준으로 도 단위로 환산하고,
# 실제 카카오맵 측정값을 통해 검증된 값입니다 (1도 ≈ 111km 기준).
bandwidth = 0.0022

# 경도(longitude)와 위도(latitude) 순서로 coords를 구성
coords = np.array([[p.x, p.y] for p in gdf['point']])

# MeanShift 모델 생성 및 학습
model = MeanShift(bandwidth=bandwidth, bin_seeding=True)
model.fit(coords)
clusters = model.labels_

gdf['cluster'] = clusters

# 클러스터 중심점 계산
optimal_locations = {}
for cluster_id in np.unique(clusters):
    if cluster_id != -1:
        center_lon, center_lat = model.cluster_centers_[cluster_id]
        optimal_locations[cluster_id] = (center_lat, center_lon)

# 시각화 부분은 기존 코드 유지
colors = plt.cm.tab10.colors
color_list = [f'#{int(r*255):02x}{int(g*255):02x}{int(b*255):02x}' for r, g, b in colors]

# 중심점 좌표 설정
if optimal_locations:
    first_cluster_id = list(optimal_locations.keys())[0]
    center_lat, center_lon = optimal_locations[first_cluster_id]
elif not gdf.empty:
    center_lat = gdf['point'].iloc[0].y
    center_lon = gdf['point'].iloc[0].x
else:
    center_lat = 35.17
    center_lon = 129.12

m = folium.Map(location=[center_lat, center_lon], zoom_start=14)

# 클러스터 포인트 시각화
for idx, row in gdf.iterrows():
    lat = row['point'].y
    lon = row['point'].x
    cluster = row['cluster']
    
    color = color_list[cluster % len(color_list)]  # 모든 포인트가 클러스터에 할당됨
    
    folium.CircleMarker(
        location=[lat, lon],
        radius=3,
        color=color,
        fill=True,
        fill_color=color,
        fill_opacity=0.8,
        popup=f'cluster: {cluster}'
    ).add_to(m)

# 최적 위치 마커 추가
for cluster_id, loc in optimal_locations.items():
    folium.Marker(
        location=[loc[0], loc[1]],
        icon=folium.Icon(color='red', icon='info-sign'),
        popup=f'Optimal Location for Cluster {cluster_id}'
    ).add_to(m)

m.save('meanshift_clusters_all_gdf.html')
```

```python
# --- MeanShift 클러스터링 및 새로운 GeoDataFrame 생성 시작 ---

# MeanShift의 대역폭(bandwidth) 설정
# 이 값은 5~9세 아동의 15분 평균 보행 거리(약 540m)를 기준으로 도 단위로 환산하고,
# 실제 카카오맵 측정값을 통해 검증된 값입니다 (1도 ≈ 111km 기준).
bandwidth = 0.0022

# 경도(longitude)와 위도(latitude) 순서로 coords를 구성합니다.
coords = np.array([[p.x, p.y] for p in gdf['point']])

# MeanShift 모델 생성 및 학습
model = MeanShift(bandwidth=bandwidth, bin_seeding=True)
model.fit(coords)

# 각 원본 데이터 포인트에 할당된 클러스터 ID
clusters = model.labels_

# 각 클러스터 ID별 데이터 포인트 수 계산
# value_counts()는 시리즈의 고유 값별 빈도수를 반환합니다.
# to_dict()를 사용하여 딕셔너리로 변환합니다.
cluster_counts = pd.Series(clusters).value_counts().to_dict()

# 새로운 GeoDataFrame을 위한 데이터 리스트 초기화
cluster_summary_data = []

# 각 클러스터의 중심점과 데이터 수를 추출하여 리스트에 추가
# model.cluster_centers_에는 클러스터 중심점들이 저장되어 있습니다.
for cluster_id, (center_lon, center_lat) in enumerate(model.cluster_centers_):
    # 만약 model.cluster_centers_가 모든 클러스터 ID를 순서대로 포함하지 않는다면,
    # np.unique(clusters)를 사용하여 실제 클러스터 ID를 반복하는 것이 더 안전합니다.
    # 여기서는 model.cluster_centers_의 인덱스가 cluster_id와 일치한다고 가정합니다.
    
    # 해당 클러스터 ID에 대한 데이터 포인트 수 가져오기
    # cluster_counts 딕셔너리에서 해당 cluster_id의 값을 찾고, 없으면 0을 반환합니다 (안전 장치).
    count = cluster_counts.get(cluster_id, 0)
    
    # 각 클러스터 정보를 딕셔너리 형태로 저장
    cluster_info = {
        'cluster_id': cluster_id,
        'data_points_count': count,
        'geometry': Point(center_lon, center_lat) # Point 객체로 저장 (경도, 위도)
    }
    cluster_summary_data.append(cluster_info)

# 리스트를 GeoDataFrame으로 변환
# crs는 원본 gdf의 crs와 동일하게 설정합니다.
clusters_gdf = gpd.GeoDataFrame(cluster_summary_data, crs=gdf.crs)

# --- 새로운 GeoDataFrame 생성 종료 ---

# 생성된 GeoDataFrame 확인
print("새로운 clusters_gdf GeoDataFrame:")
print(clusters_gdf.head())

# 모든 클러스터의 정보 출력 (선택 사항)
print("\n--- 모든 클러스터의 요약 정보 ---")
print(clusters_gdf[['cluster_id', 'data_points_count', 'geometry']])
```
